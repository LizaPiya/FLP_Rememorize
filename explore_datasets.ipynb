```python
# Change DATASET_ID to explore any dataset
DATASET_ID = ""   # e.g. "cnn_dailymail"
CONFIG     = None # e.g. "3.0.0" — set to None if not needed

ds = load_dataset(DATASET_ID, CONFIG) if CONFIG else load_dataset(DATASET_ID)
split_name = next(iter(ds.keys()))
df = ds[split_name].to_pandas()
print(f"Split: {split_name}  |  Rows: {len(df):,}  |  Columns: {list(df.columns)}")
df.head()
```

```python
# Token length distribution (useful for checking context window fit)
TEXT_COL = df.columns[0]   # change to the column with the main text

df['token_len'] = df[TEXT_COL].str.split().str.len()
df['token_len'].describe(percentiles=[0.25, 0.5, 0.75, 0.9, 0.95, 0.99])
```
